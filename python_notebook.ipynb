```python
import pandas as pd
import numpy as np

file_name = "Cleaned_Sample_Dataset.xlsx"

df = pd.read_excel(file_name)

df["Event_Date"] = pd.to_datetime(df["Event_Date"], errors="coerce")

df["Approval_Rate"] = np.where(
    df["Requests"] > 0,
    (df["Successful_Requests"] / df["Requests"]) * 100,
    0
)

df["Failure_Rate"] = np.where(
    df["Requests"] > 0,
    (df["Failed_Requests"] / df["Requests"]) * 100,
    0
)

df["Profit"] = df["Revenue"] - df["Cost"]

df["Profit_Margin"] = np.where(
    df["Revenue"] > 0,
    (df["Profit"] / df["Revenue"]) * 100,
    0
)

df["Revenue_Per_Request"] = np.where(
    df["Successful_Requests"] > 0,
    df["Revenue"] / df["Successful_Requests"],
    0
)

df["Cost_Per_Request"] = np.where(
    df["Successful_Requests"] > 0,
    df["Cost"] / df["Successful_Requests"],
    0
)

df["Launch_Month"] = df["Event_Date"].dt.to_period("M").astype(str)


# 1. Overall Business KPIs

total_users = df["User_ID"].nunique()
total_requests = df["Requests"].sum()
successful_requests = df["Successful_Requests"].sum()
failed_requests = df["Failed_Requests"].sum()

total_revenue = df["Revenue"].sum()
total_cost = df["Cost"].sum()
total_profit = total_revenue - total_cost

overall_approval_rate = (
    successful_requests / total_requests * 100
    if total_requests > 0 else 0
)

overall_failure_rate = (
    failed_requests / total_requests * 100
    if total_requests > 0 else 0
)

overall_profit_margin = (
    total_profit / total_revenue * 100
    if total_revenue > 0 else 0
)

average_latency = df["Latency_ms"].mean()
average_recovery_time = df["Recovery_Time_min"].mean()
average_quality_score = df["Quality_Score"].mean()

compliance_rate = (
    (df["Compliance_Status"] == "Compliant").mean() * 100
)

rollback_rate = (
    (df["Rollback_Triggered"] == "Yes").mean() * 100
)


# 2. Overall KPI Table

overall_kpi = pd.DataFrame({
    "Metric": [
        "Total Users",
        "Total Requests",
        "Successful Requests",
        "Failed Requests",
        "Total Revenue",
        "Total Cost",
        "Total Profit",
        "Approval Rate",
        "Failure Rate",
        "Profit Margin",
        "Average Latency (ms)",
        "Average Recovery Time (min)",
        "Average Quality Score",
        "Compliance Rate",
        "Rollback Rate"
    ],
    "Value": [
        total_users,
        total_requests,
        successful_requests,
        failed_requests,
        total_revenue,
        total_cost,
        total_profit,
        overall_approval_rate,
        overall_failure_rate,
        overall_profit_margin,
        average_latency,
        average_recovery_time,
        average_quality_score,
        compliance_rate,
        rollback_rate
    ]
})


# 3. Monthly Performance

monthly_performance = df.groupby("Launch_Month").agg(
    Users=("User_ID", "nunique"),
    Requests=("Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Average_Latency_ms=("Latency_ms", "mean"),
    Average_Recovery_Time_min=("Recovery_Time_min", "mean"),
    Average_Quality_Score=("Quality_Score", "mean")
).reset_index()

monthly_performance["Approval_Rate"] = (
    monthly_performance["Successful_Requests"]
    / monthly_performance["Requests"] * 100
)

monthly_performance["Failure_Rate"] = (
    monthly_performance["Failed_Requests"]
    / monthly_performance["Requests"] * 100
)

monthly_performance["Profit_Margin"] = (
    monthly_performance["Profit"]
    / monthly_performance["Revenue"] * 100
)


# 4. Region Performance

region_performance = df.groupby("Region").agg(
    Users=("User_ID", "nunique"),
    Requests=("Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Average_Latency_ms=("Latency_ms", "mean"),
    Average_Quality_Score=("Quality_Score", "mean"),
    Average_Recovery_Time_min=("Recovery_Time_min", "mean")
).reset_index()

region_performance["Approval_Rate"] = (
    region_performance["Successful_Requests"]
    / region_performance["Requests"] * 100
)

region_performance["Failure_Rate"] = (
    region_performance["Failed_Requests"]
    / region_performance["Requests"] * 100
)

region_performance["Profit_Margin"] = (
    region_performance["Profit"]
    / region_performance["Revenue"] * 100
)


# 5. Platform Performance

platform_performance = df.groupby("Platform").agg(
    Users=("User_ID", "nunique"),
    Requests=("Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Average_Latency_ms=("Latency_ms", "mean"),
    Average_Quality_Score=("Quality_Score", "mean"),
    Average_Recovery_Time_min=("Recovery_Time_min", "mean")
).reset_index()

platform_performance["Approval_Rate"] = (
    platform_performance["Successful_Requests"]
    / platform_performance["Requests"] * 100
)

platform_performance["Failure_Rate"] = (
    platform_performance["Failed_Requests"]
    / platform_performance["Requests"] * 100
)

platform_performance["Profit_Margin"] = (
    platform_performance["Profit"]
    / platform_performance["Revenue"] * 100
)


# 6. Version Performance

version_performance = df.groupby("Version").agg(
    Requests=("Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Average_Latency_ms=("Latency_ms", "mean"),
    Average_Quality_Score=("Quality_Score", "mean"),
    Average_Recovery_Time_min=("Recovery_Time_min", "mean")
).reset_index()

version_performance["Approval_Rate"] = (
    version_performance["Successful_Requests"]
    / version_performance["Requests"] * 100
)

version_performance["Failure_Rate"] = (
    version_performance["Failed_Requests"]
    / version_performance["Requests"] * 100
)

version_performance["Profit_Margin"] = (
    version_performance["Profit"]
    / version_performance["Revenue"] * 100
)


# 7. Launch Status Analysis

launch_status = df.groupby("Launch_Status").agg(
    Records=("Launch_Status", "count"),
    Requests=("Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Average_Quality_Score=("Quality_Score", "mean"),
    Average_Latency_ms=("Latency_ms", "mean")
).reset_index()

launch_status["Approval_Rate"] = (
    launch_status["Successful_Requests"]
    / launch_status["Requests"] * 100
)

launch_status["Failure_Rate"] = (
    launch_status["Failed_Requests"]
    / launch_status["Requests"] * 100
)


# 8. Compliance Analysis

compliance_analysis = df.groupby("Compliance_Status").agg(
    Records=("Compliance_Status", "count"),
    Requests=("Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum")
).reset_index()

compliance_analysis["Record_Share"] = (
    compliance_analysis["Records"]
    / compliance_analysis["Records"].sum() * 100
)


# 9. Rollback Analysis

rollback_analysis = df.groupby("Rollback_Triggered").agg(
    Records=("Rollback_Triggered", "count"),
    Requests=("Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Average_Latency_ms=("Latency_ms", "mean"),
    Average_Recovery_Time_min=("Recovery_Time_min", "mean"),
    Average_Quality_Score=("Quality_Score", "mean")
).reset_index()

rollback_analysis["Failure_Rate"] = (
    rollback_analysis["Failed_Requests"]
    / rollback_analysis["Requests"] * 100
)


# 10. Data Quality Summary

missing_values = df.isnull().sum().reset_index()

missing_values.columns = [
    "Column",
    "Missing_Values"
]

missing_values["Missing_Percentage"] = (
    missing_values["Missing_Values"]
    / len(df) * 100
)

duplicate_count = df.duplicated().sum()

data_quality = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Duplicate Rows",
        "Total Missing Values"
    ],
    "Value": [
        len(df),
        len(df.columns),
        duplicate_count,
        df.isnull().sum().sum()
    ]
})


# 11. Unit Economics

unit_economics = pd.DataFrame({
    "Metric": [
        "Revenue Per Successful Request",
        "Cost Per Successful Request",
        "Profit Per Successful Request",
        "Overall Profit Margin"
    ],
    "Value": [
        total_revenue / successful_requests
        if successful_requests > 0 else 0,

        total_cost / successful_requests
        if successful_requests > 0 else 0,

        total_profit / successful_requests
        if successful_requests > 0 else 0,

        overall_profit_margin
    ]
})


# 12. Top Performing Regions

top_regions = region_performance.sort_values(
    "Profit",
    ascending=False
).head(5)


# 13. Highest Risk Regions

risk_regions = region_performance.sort_values(
    "Failure_Rate",
    ascending=False
).head(5)


# 14. Platform Risk Analysis

platform_risk = platform_performance.sort_values(
    "Failure_Rate",
    ascending=False
)


# 15. Quality and Reliability Summary

reliability_summary = pd.DataFrame({
    "Metric": [
        "Average Latency (ms)",
        "Average Recovery Time (min)",
        "Average Quality Score",
        "Compliance Rate",
        "Rollback Rate"
    ],
    "Value": [
        average_latency,
        average_recovery_time,
        average_quality_score,
        compliance_rate,
        rollback_rate
    ]
})


# 16. Business Insights

best_region = region_performance.loc[
    region_performance["Profit"].idxmax(),
    "Region"
]

best_platform = platform_performance.loc[
    platform_performance["Profit"].idxmax(),
    "Platform"
]

highest_risk_region = region_performance.loc[
    region_performance["Failure_Rate"].idxmax(),
    "Region"
]

highest_risk_platform = platform_performance.loc[
    platform_performance["Failure_Rate"].idxmax(),
    "Platform"
]

best_version = version_performance.loc[
    version_performance["Profit"].idxmax(),
    "Version"
]

insights = pd.DataFrame({
    "Business_Insight": [
        f"{best_region} generated the highest total profit.",
        f"{best_platform} generated the highest platform-level profit.",
        f"{highest_risk_region} had the highest failure rate among regions.",
        f"{highest_risk_platform} had the highest failure rate among platforms.",
        f"{best_version} generated the highest total profit among versions.",
        f"Overall approval rate was {overall_approval_rate:.2f}%.",
        f"Overall failure rate was {overall_failure_rate:.2f}%.",
        f"Overall profit margin was {overall_profit_margin:.2f}%.",
        f"Compliance rate was {compliance_rate:.2f}%.",
        f"Rollback trigger rate was {rollback_rate:.2f}%."
    ]
})


# 17. Export Everything into One Excel File

output_file = "Python Analysis Result.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

    overall_kpi.to_excel(
        writer,
        sheet_name="Overall KPI",
        index=False
    )

    monthly_performance.to_excel(
        writer,
        sheet_name="Monthly Performance",
        index=False
    )

    region_performance.to_excel(
        writer,
        sheet_name="Region Performance",
        index=False
    )

    platform_performance.to_excel(
        writer,
        sheet_name="Platform Performance",
        index=False
    )

    version_performance.to_excel(
        writer,
        sheet_name="Version Performance",
        index=False
    )

    launch_status.to_excel(
        writer,
        sheet_name="Launch Status",
        index=False
    )

    compliance_analysis.to_excel(
        writer,
        sheet_name="Compliance",
        index=False
    )

    rollback_analysis.to_excel(
        writer,
        sheet_name="Rollback Analysis",
        index=False
    )

    data_quality.to_excel(
        writer,
        sheet_name="Data Quality",
        index=False
    )

    missing_values.to_excel(
        writer,
        sheet_name="Missing Values",
        index=False
    )

    unit_economics.to_excel(
        writer,
        sheet_name="Unit Economics",
        index=False
    )

    top_regions.to_excel(
        writer,
        sheet_name="Top Regions",
        index=False
    )

    risk_regions.to_excel(
        writer,
        sheet_name="Risk Regions",
        index=False
    )

    platform_risk.to_excel(
        writer,
        sheet_name="Platform Risk",
        index=False
    )

    reliability_summary.to_excel(
        writer,
        sheet_name="Reliability",
        index=False
    )

    insights.to_excel(
        writer,
        sheet_name="Business Insights",
        index=False
    )


print("Analysis Completed Successfully")
print("Output File:", output_file)
print("Total Rows:", len(df))
print("Total Columns:", len(df.columns))
print("Total Revenue:", round(total_revenue, 2))
print("Total Profit:", round(total_profit, 2))
print("Approval Rate:", round(overall_approval_rate, 2), "%")
print("Failure Rate:", round(overall_failure_rate, 2), "%")
print("Compliance Rate:", round(compliance_rate, 2), "%")
print("Rollback Rate:", round(rollback_rate, 2), "%")
print("\nOnly one analysis output file was created:")
print(output_file)
```
